### Q6. Median value of horsepower

```python
# 1. Mediana antes de rellenar
mediana_antes = df['horsepower'].median()

# 2. Valor más frecuente (la moda). mode() devuelve una lista, por eso el [0]
moda = df['horsepower'].mode()[0]

# 3. Rellenar los nulos con la moda
df['horsepower'] = df['horsepower'].fillna(moda)

# 4. Mediana después
mediana_despues = df['horsepower'].median()

print(mediana_antes, moda, mediana_despues)
```

### Q7. Sum of weights

```python
# Coches de Asia
asia = df[df['origin'] == 'Asia']

# Solo dos columnas y las 7 primeras filas, como array de NumPy
X = asia[['vehicle_weight', 'model_year']].head(7).values

# X traspuesta por X
XTX = X.T @ X

# Invertir
XTX_inv = np.linalg.inv(XTX)

# Vector y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# w = (XTX)^-1 · X^T · y
w = XTX_inv @ X.T @ y

w.sum()
```
